## Part 1. 리스트와 반복문

```python
scores = [88, 92, 45, 71, 100, 63, 79]
high_score_count = 0

for score in scores:
    print(f'점수: {score}')
    if score >= 80:
        print(f'  -> 80점 이상: {score}')
        high_score_count += 1

print(f'80점 이상인 점수는 총 {high_score_count}개입니다.')
```

## Part 2. 딕셔너리를 이용한 단어 빈도 분석

`split()`만 쓰면 '파이썬'과 '파이썬을'이 서로 다른 단어로 집계된다. 그래서 이 예문에 나온 조사만 제거해서 같은 단어로 세었다. `collections.Counter`는 사용하지 않았다.

```python
sentence = '''
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
'''

particles = ('에는', '에도', '으로', '에서', '까지', '은', '는', '이', '가', '을', '를')
words = sentence.split()
counts = {}

for raw_word in words:
    word = raw_word
    for particle in particles:
        if word.endswith(particle) and len(word) > len(particle):
            word = word[:-len(particle)]
            break
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1

print('전체 단어 빈도')
for word, frequency in counts.items():
    print(f'{word} : {frequency}회')

print('\n2회 이상 등장한 단어')
for word, frequency in counts.items():
    if frequency >= 2:
        print(f'{word} : {frequency}회')
```

## Part 3. 함수로 반복되는 코드 정리하기

```python
def normalize_word(word):
    particles = ('에는', '에도', '으로', '에서', '까지', '은', '는', '이', '가', '을', '를')
    for particle in particles:
        if word.endswith(particle) and len(word) > len(particle):
            return word[:-len(particle)]
    return word


def count_words(sentence):
    result = {}
    for raw_word in sentence.split():
        word = normalize_word(raw_word)
        if word in result:
            result[word] += 1
        else:
            result[word] = 1
    return result


def print_frequent_words(counts, n):
    found = False
    for word, frequency in counts.items():
        if frequency >= n:
            print(f'{word} : {frequency}회')
            found = True
    if not found:
        print(f'{n}회 이상 등장한 단어가 없습니다.')


word_counts = count_words(sentence)
print_frequent_words(word_counts, 2)
```

## Part 4. 오류 확인 및 코드 검증

```python
try:
    user_input = input('기준 횟수를 입력하세요(기본값 2): ').strip()
    standard = int(user_input) if user_input else 2
    if standard < 1:
        raise ValueError('기준 횟수는 1 이상이어야 합니다.')
except ValueError as error:
    print(f'입력 오류: {error} 기본값 2를 사용합니다.')
    standard = 2

print_frequent_words(word_counts, standard)

search_word = input('검색할 단어를 입력하세요(기본값 파이썬): ').strip() or '파이썬'
try:
    print(f'{search_word} : {word_counts[search_word]}회')
except KeyError:
    print(f"'{search_word}'은 문장에 없는 단어입니다.")
```

### 오류 해결 기록

- 발생한 오류: 숫자가 아닌 기준을 `int()`로 바꿀 때 `ValueError`가 발생했다.
- 발생 위치: 사용자 입력을 정수로 바꾸는 부분이다.
- 확인 내용: `input()`의 결과는 숫자를 입력해도 문자열이다.
- 수정 내용: 변환 부분을 `try-except`로 처리하고 잘못된 입력에는 기본값 2를 사용했다.

## 도전 문제 - 가장 많이 등장한 단어 찾기

```python
max_frequency = max(word_counts.values())
for word, frequency in word_counts.items():
    if frequency == max_frequency:
        print(f'가장 많이 등장한 단어: {word} ({frequency}회)')
```
